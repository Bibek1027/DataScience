# Loops

### For loop

```python
for i in range(1, 11):
    print(i)
```

```python
for i in range(2,21,2):
    print(i)
```

```python
total = 0

for i in range(1,101):
    total += i
print("Sum", total)
```

```python
num = int(input("Enter a number: "))

for i in range(1,11):
    print(num, "x", i, "=", num * i)
```

```python
n = int(input("Enter a number"))
count = 0

for i in range(1,n):
    if i % 3 == 0:
        count += 1

print(count)
```

### While Loop

```python
num = int(input("Enter a number: "))
count = 0

while num > 0:
    num = num // 10
    count += 1

print("Number of digits:", count)
```

```python
num = int(input("Enter a number: "))
reverse = 0

while num > 0:
    a = num % 10
    reverse = reverse * 10 + a
    num = num //10 

print('Reverse',reverse)
```

# Strings

```python
word = input("Enter a word: ")

print("First character:" ,word[0])
print("Last character:" ,word[-1])
print("Length", len(word))
```

```python
word = input("Enter a word: ")

print('Reverse:', word[::-1])
```

```python
word = input("Enter a word: ")
character = input("Enter a character: ")

count = word.count(character)

print(character, 'appears', count, 'times')
```

```python
word = input("Enter a word: ")
character = input("Enter a character: ")
count = 0

for i in word:
    if i == character:
        count += 1

print("Count:", count)
```

```python
word = "Python"

for character in word:
    print(character)
```

# Lists

```python
numbers = [10, 25, 7, 40, 15, 30]
total = 0

for i in numbers:
    total += i

print("Total:", total)
```

```python
numbers = [10, 25, 7, 40, 15, 30]
large = 0

for i in numbers:
    if large < i:
        large = i

print('Largest:', large)
```

```python
numbers = [10, 25, 7, 40, 15, 30, 8, 55]

for i in numbers:
    if i > 20:
        print(i)
```

```python
numbers = [10, 25, 7, 40, 15, 30, 8, 55]
even = 0
odd = 0

for i in numbers:
    if i % 2 == 0:
        even += 1
    else:
        odd += 1

print('Even:', even)
print("Odd:", odd)
```

# Dictionaries

```python
student = {
    "name": "Alex",
    "age": 21,
    "python_score": 88
}

print("Name:", student["name"])
print("Age:", student['age'])
print("Score:", student['python_score'])
```

```python
student = {
    "name": "Alex",
    "age": 21,
    "python_score": 88
}

student["python_score"] = 95
student["passed"] = True

print(student)
```

```python
student = {
    "name": "Alex",
    "age": 21,
    "python_score": 88,
    "passed": True
}

for key, value in student.items():
    #print(key + ":", value)
    print(f"{key}: {value}")
```

```python
students = {
    "Alex": 85,
    "John": 92,
    "Sarah": 78,
    "Mike": 95,
    "Emma": 88
}

highest_score = 0
highest_student = ""

for key, value in students.items():
    if value > highest_score:
        highest_score = value
        highest_student = key

print("Highest Score:", highest_student)
print("Score:", highest_score)
```

# Functions

```python
def calculate_square(num):
    return num ** 2

result = calculate_square(5)
print("Square:", result)
```

```python
def add_numbers(a, b):
    return a + b

result = add_numbers(10, 20)
print(result)
```

```python
def calculate_average(a,b,c):
    return (a+b+c)/3

result = calculate_average(80, 90, 70)
print(result)
```

```python
def find_largest(numbers):
    great = 0

    for i in numbers:
        if i > great:
            great = i

    return great

numbers = [10, 25, 7, 40, 15]

result = find_largest(numbers)
print(result)
```
