```python
class Car:
    def __init__(self, brand, model, year):
        self.brand = brand
        self.model = model
        self.year = year

car1= Car("Toyota", "Corolla", 2020)
car2= Car("Honda", "Civic", 2022)

print(car1.brand,car1.model, car1.year)
print(car2.brand, car2.model, car2.year)
```

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        return f"Hello, my name is {self.name} and I am {self.age} years old."


s1 = Student("Alex", 21)

print(s1.introduce())
```

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def check_result(self):
        if self.marks >= 40:
            print(self.name, "passed")
        else:
            print(self.name, "failed")

s1 = Student("Bibek", 80)
s1.check_result()

s2 = Student("Alex", 39)
s2.check_result()
```

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance = self.balance + amount


own1 = BankAccount("Bibek", 5000)

print("Curent Balance:",own1.balance)

own1.deposit(2000)

print("New Balance:",own1.balance)
```

```python
#Main code

class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display_result(self):
        total = 0

        for mark in self.marks:
            total += mark

        average = total / len(self.marks)

        print("Name:", self.name)
        print("Average:", average)

        if average >= 40:
            print("Result: Pass")
        else:
            print("Result: Fail")

        print()


students = []

try:
    with open("students.txt", "r") as file:
        data = file.readlines()

    for student in data:
        pieces = student.strip().split(",")

        name = pieces[0]

        marks = []
        for mark in pieces[1:]:
            marks.append(int(mark))

        student = Student(name, marks)
        students.append(student)

except FileNotFoundError:
    print("File not found.")


for student in students:
    student.display_result()
```

```python
try:
    with open('students.txt', 'r') as file:
        data = file.readlines()

    class Student:
        def __init__(self, name, marks):
            self.name = name
            self.marks = marks

        def display_result(self):
            total = 0

            for mark in self.marks:
                total += mark

            avg = total / len(self.marks)

            print("Name:", self.name)
            print("Average marks:", avg)

            if avg >= 40:
                print("Result: Pass")
            else:
                print("Result: Fail")

    for student in data:
        parts = student.strip().split(",")

        name = parts[0]

        marks = []
        for mark in parts[1:]:
            marks.append(int(mark))

        s = Student(name, marks)
        s.display_result()

except FileNotFoundError:
    print("File not found.")
```
