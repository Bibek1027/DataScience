```python
num = int(input("Enter a number:"))

negative = num < 0

num = abs(num)
rev = 0

while num > 0:
    dig = num % 10
    rev = rev * 10 + dig
    num = num // 10

if negative:
    rev = -rev

print("Reverse:", rev)
```

```python
def count_even(numbers):
    count = 0

    for number in numbers:
        if number % 2 == 0:
            count += 1

    return count


numbers = [2, 5, 8, 11, 14]

result = count_even(numbers)

print(result)
```

```python
def find_smallest(numbers):
    smallest = numbers[0]

    for number in numbers:
        if number < smallest:
            smallest = number

    return smallest


numbers = [12, 5, 8, 2, 19]

result = find_smallest(numbers)

print(result)
```

```python
students = {
    "Bibek": [80, 70, 90],
    "Alex": [35, 45, 40],
    "Sarah": [90, 85, 95],
    "John": [60, 55, 65]
}

highest_average = 0
highest_student = ""

for name, marks in students.items():
    total = 0

    for mark in marks:
        total += mark

    average = total / len(marks)

    if average >= 40:
        status = "Pass"
    else:
        status = "Fail"

    print(name, "- Average:", average, "-", status)

    if average > highest_average:
        highest_average = average
        highest_student = name

print()
print("Highest Average:", highest_student)
print("Average:", highest_average)
```

```python
cart = [
    {"name": "Laptop", "price": 800, "quantity": 1},
    {"name": "Mouse", "price": 25, "quantity": 2},
    {"name": "Keyboard", "price": 50, "quantity": 1}
]

subtotal = 0

for item in cart:
    total = item["price"] * item["quantity"]
    print(item["name"] + ":", total)

    subtotal += total

if subtotal >= 1000:
    discount = subtotal * 0.10
else:
    discount = 0

final_total = subtotal - discount

print()
print("Subtotal:", subtotal)
print("Discount:", discount)
print("Final Total:", final_total)
```

```python
employees = [
    {"name": "Bibek", "salary": 50000},
    {"name": "Alex", "salary": 42000},
    {"name": "Sarah", "salary": 65000},
    {"name": "John", "salary": 38000}
]


def analyze_salaries(employees):
    highest_salary = employees[0]["salary"]
    highest_name = employees[0]["name"]

    lowest_salary = employees[0]["salary"]
    lowest_name = employees[0]["name"]

    total_salary = 0

    for employee in employees:
        name = employee["name"]
        salary = employee["salary"]

        total_salary += salary

        if salary > highest_salary:
            highest_salary = salary
            highest_name = name

        if salary < lowest_salary:
            lowest_salary = salary
            lowest_name = name

    average_salary = total_salary / len(employees)

    return highest_name, lowest_name, average_salary


highest, lowest, average = analyze_salaries(employees)

print("Highest paid:", highest)
print("Lowest paid:", lowest)
print("Average salary:", average)
```

```python
employees = [
    {"name": "Bibek", "salary": 50000},
    {"name": "Alex", "salary": 42000},
    {"name": "Sarah", "salary": 65000},
    {"name": "John", "salary": 38000}
]


def analyze_salaries(employees):
    highest = employees[0]
    lowest = employees[0]
    total_salary = 0

    for employee in employees:
        salary = employee["salary"]

        total_salary += salary

        if salary > highest["salary"]:
            highest = employee

        if salary < lowest["salary"]:
            lowest = employee

    average_salary = total_salary / len(employees)

    return highest["name"], lowest["name"], average_salary


highest, lowest, average = analyze_salaries(employees)

print("Highest paid:", highest)
print("Lowest paid:", lowest)
print("Average salary:", average)
```
