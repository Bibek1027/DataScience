# Operators

### Arithmetic Operators

```python
item1 = 250
item2 = 100

total = (item1*3) + (item2*2)
print("Total Cost:", total)
average = total /(3+2)
print("Average cost per item", average)
```

```python
sec = int(input("Enter a number of seconds"))
hour= sec // 3600
remaining = sec % 3600
mins = remaining // 60
secs = remaining % 60

print("Hours:", hour)
print('Minutes:', mins)
print('Seconds', secs)
```

### Comparison Operators

```python
num1 = int(input("Enter 1st nbr"))
num2 = int(input("Enter 2nd nbr"))

print("First number is greater", num1 > num2)
print("Second number is greater", num1 < num2)
print("Both numbers are equal", num1 == num2)
```

```python
age = int(input("Enter your age"))

print('Under 13', age < 13)
print('Teenager', age > 13 and age <18)
print("Adult", age > 18)
```

### Logical Operators

```python
username = "admin"
password = "admin123"

user = input("enter your username")
password2 = input("enter password")

print('Access Granted', user == username and password == password2 )
print('Access Denied', not (user == username and password == password2))
```

```python
age = int(input("enter age"))
pur_amt = int(input("enter purchase amount"))

print('You are eligible for a discount', age >= 60 or pur_amt >= 5000)
print('You are not eligible for a discount', not (age >= 60 or pur_amt >= 5000))
```

## Challenges

```python
name = input("Enter student's name: ")
age = int(input("Enter age: "))

sub1 = float(input("Enter mark for subject 1: "))
sub2 = float(input("Enter mark for subject 2: "))
sub3 = float(input("Enter mark for subject 3: "))

total = sub1 + sub2 + sub3
average = total / 3

if sub1 >= 40 and sub2 >= 40 and sub3 >= 40:
    result = "Pass"

    if average >= 90:
        grade = "A"
    elif average >= 80:
        grade = "B"
    elif average >= 70:
        grade = "C"
    elif average >= 60:
        grade = "D"
    else:
        grade = "F"
else:
    result = "Fail"
    grade = "N/A"

if age < 13:
    age_category = "Under 13"
elif age < 18:
    age_category = "Teenager"
else:
    age_category = "Adult"

print("Student:", name)
print("Total:", total)
print("Average:", average)
print("Result:", result)
print("Grade:", grade)
print("Age Category:", age_category)
```

```python
name = input("Enter customer's name: ")
age = int(input("Enter age: "))

item_price = float(input("Enter price: "))
item_qty = int(input("Enter quantity"))

subtotal = item_price * item_qty

if age >= 60:
    dis = (20/100)*subtotal
    final_price = subtotal - dis

elif subtotal >= 5000:
    dis = (10/100)* subtotal
    final_price = subtotal - dis

else: 
    dis = 0
    final_price = subtotal

print("Customer:", name)
print("Subtotal:", subtotal)
print("Discount:", dis)
print("Final Amount:", final_price)
```
